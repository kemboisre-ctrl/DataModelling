### dimcustomers


```python
%sql
CREATE OR REPLACE TABLE datamodeling.gold.DimCustomers
SELECT *,
row_number() OVER (ORDER BY order_customer_id) as DimCustomerKey
FROM
( 
SELECT 
DISTINCT (order_customer_id),
customer_email,
customer_name
 FROM datamodeling.silver.silver_table)
```

```python
%sql
CREATE OR REPLACE TABLE datamodeling.gold.DimProducts
SELECT *,
row_number() OVER (ORDER BY product_id) as DimProductKey
FROM
( 
SELECT
DISTINCT (product_id),
product_name,
product_category
 FROM datamodeling.silver.silver_table)
```

```python
%sql
SELECT * FROM datamodeling.gold.DimProducts
```

DIM PAYMENTS

```python
%sql
CREATE OR REPLACE TABLE datamodeling.gold.DimPayments
SELECT *,
row_number() OVER (ORDER BY payment_type) as DimPaymentKey
FROM
( 
SELECT
DISTINCT (payment_type)
 FROM datamodeling.silver.silver_table)
```

```python
%sql
SELECT * FROM datamodeling.gold.DimPayments
```

DIM SALES

```python
%sql
CREATE OR REPLACE TABLE datamodeling.gold.DimSales
AS
SELECT 
row_number() OVER (ORDER BY order_id) as DimSalesKey,
order_id,
 order_date,
 order_customer_id,
 customer_name,
 order_status,
 customer_email,
 order_address,
 product_id,
 product_name,
 product_category,
 revenue,
 payment_type,
 country,
 last_updated,
 US,
 process_date
 FROM datamodeling.silver.silver_table
```

```python
%sql
SELECT * FROM datamodeling.gold.DimSales
```

FACT TABLE

```python
%sql
CREATE OR REPLACE TABLE datamodeling.gold.FactSales
SELECT
S.dimSalesKey,
C.dimCustomerKey,
P.dimProductKey,
PY.dimPaymentKey,
F.quantity,
F.unit_price
FROM
datamodeling.silver.silver_table F
LEFT JOIN datamodeling.gold.DimSales S
ON F.order_id = S.order_id
LEFt JOIN datamodeling.gold.DimCustomers C
ON F.order_customer_id = C.order_customer_id
LEFT JOIN datamodeling.gold.DimProducts P
ON F.product_id = P.product_id
LEFT JOIN datamodeling.gold.DimPayments PY
```

```python
%sql
SELECT * FROM datamodeling.gold.FactSales
```
