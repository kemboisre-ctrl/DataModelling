```sql
%python
if spark.catalog.tableExists("datamodeling2.Bronze.bronze_table"):
    last_load_date = spark.sql("""
        SELECT max(order_date)
        FROM datamodeling2.Bronze.bronze_table
    """).collect()[0][0]
else:
    last_load_date = "1970-01-01"
```

```sql
%python
last_load_date
```

```sql
%python

spark.sql(f"""SELECT * FROM datamodeling2.default.source_data
         WHERE order_date > '{last_load_date}'""").createOrReplaceTempView ("datamodeling2.Bronze.bronze_source")
```

```sql
%sql
CREATE OR REPLACE TABLE datamodeling2.Bronze.bronze_source AS
SELECT * FROM bronze_source
```
